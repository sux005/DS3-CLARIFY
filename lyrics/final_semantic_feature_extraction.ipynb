```python
import pandas as pd
import os
import time
import requests
import spotipy
from dotenv import load_dotenv
from sentence_transformers import SentenceTransformer
from spotipy.oauth2 import SpotifyClientCredentials
pd.set_option('display.max_columns', None)
```

```python
load_dotenv()
client_id = os.getenv('SPOTIPY_CLIENT_ID')
client_secret = os.getenv('SPOTIPY_CLIENT_SECRET')
sp = spotipy.Spotify(auth_manager=SpotifyClientCredentials(
    client_id=client_id, 
    client_secret=client_secret
))
```

```python
semantic_data = pd.read_csv('C:/Users/maxph/OneDrive/Desktop/DS3-CLARIFY/data/training_data/processed_lyrics_features.csv')
semantic_data
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Generate embeddings for your clean lyrics column
# Ensure your lyrics column has no missing values (NaN) before running this
lyrics_list = semantic_data['Clean_Lyrics'].fillna("").tolist()
print("Generating embeddings... this might take a moment.")
embeddings = model.encode(lyrics_list, show_progress_bar=True)

# 4. Create columns for the embeddings in a new DataFrame
# The MiniLM model outputs a 384-dimensional vector, creating BERT_Embedding_0 to BERT_Embedding_383
embedding_data = pd.DataFrame(
    embeddings, 
    columns=[f'BERT_Embedding_{i}' for i in range(embeddings.shape[1])]
)

# 5. Concatenate the original features with the new embedding features
final_data = pd.concat([semantic_data, embedding_data], axis=1)
final_data
```

```python
feature_data = pd.read_csv('C:/Users/maxph/OneDrive/Desktop/DS3-CLARIFY/data/training_data/theme_lyrics_features.csv')
feature_data
```

```python
def fetch_spotify_id(row):
    # Convert to string in case of missing data (NaN)
    song = str(row['Song Title']) 
    artist = str(row['Artist'])
    
    # Format the query specifically for the Spotify search engine
    query = f"track:{song} artist:{artist}"
    
    try:
        # Search for 1 track matching the query
        results = sp.search(q=query, type='track', limit=1)
        items = results['tracks']['items']
        
        # If the list is not empty, extract the ID
        if items:
            return items[0]['id']
        else:
            print(f"No match found for: {song} by {artist}")
            return None
            
    except Exception as e:
        print(f"API Error fetching {song}: {e}")
        return None
```

```python
feature_data['SONG_ID'] = feature_data.apply(fetch_spotify_id, axis=1)
feature_data
```

```python
final_CBM_input_data = pd.DataFrame()
final_CBM_input_data['SONG_TITLE'] = feature_data['Song Title']
final_CBM_input_data['ARTIST_NAME'] = feature_data['Artist']
final_CBM_input_data['SONG_ID'] = feature_data['SONG_ID']
feature_data = feature_data.drop(['Song Title','Artist','Release_Year', 'Reviewer','SONG_ID'], axis=1)
final_data = final_data.drop(['SONG_ID', 'Clean_Lyrics'], axis=1)
final_CBM_input_data = pd.concat([final_CBM_input_data, final_data], axis=1)
final_CBM_input_data
```
